```python
# Step 6: Train the model
from scripts.train import train
model, log = train()
```

```python
# Step 7: View results
print(f"Test F1: {log['test_metrics'].get('test_f1', 0):.4f}")
print(f"Test Precision: {log['test_metrics'].get('test_precision', 0):.4f}")
print(f"Test Recall: {log['test_metrics'].get('test_recall', 0):.4f}")
print(f"Training time: {log['training_time_min']:.1f} min")
```

```python
# Step 8: Save the trained model to Google Drive
import shutil

src_model = '/content/viora-ner/models/checkpoints/best_model'
dst_model = '/content/drive/MyDrive/viora_ner_models/best_model'

# Remove old model if exists
if os.path.exists(dst_model):
    shutil.rmtree(dst_model)

# Copy to Drive
shutil.copytree(src_model, dst_model)
print(f'Model saved to Google Drive: {dst_model}')

# Show saved files
for f in sorted(os.listdir(dst_model)):
    size_mb = os.path.getsize(os.path.join(dst_model, f)) / 1024 / 1024
    print(f'  {f} ({size_mb:.1f} MB)')
```
